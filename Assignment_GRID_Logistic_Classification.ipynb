```python
import pandas as pd
dataset=pd.read_csv("CKD.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
dataset["classification_yes"].value_counts()
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bp', 'al', 'su', 'bgr', 'bu', 'sc', 'sod', 'pot', 'hrmo', 'pcv','wc', 'rc', 'sg_b', 'sg_c', 'sg_d', 'sg_e', 'rbc_normal', 'pc_normal','pcc_present', 'ba_present', 'htn_yes', 'dm_yes', 'cad_yes','appet_yes', 'pe_yes', 'ane_yes']]
dependent=dataset[['classification_yes']]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(independent,dependent,test_size=1/3,random_state=0)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
param_grid={'solver':['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga'],'penalty':['l2']}
grid=GridSearchCV(LogisticRegression(),param_grid,refit=True,verbose=3,n_jobs=-1,scoring='f1_weighted')
grid=grid.fit(X_train,y_train)
```

```python
grid_pred=grid.predict(X_test)
re=grid.cv_results_
rscore=grid.best_params_
print(f"The R_score value for best parameter:{rscore}")
```

```python
table=pd.DataFrame.from_dict(re)
table
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,grid_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report
clf_report=classification_report(y_test,grid_pred)
print(clf_report)
```

```python
from sklearn.metrics import roc_auc_score   ##roc_auc - receiver.operating.curve_area.under.curve
roc_auc_score(y_test,grid.predict_proba(X_test)[:,1])  ## grid.predict_proba - probability outputs from your tuned model (inside GridSearchCV)
                                                        ## [:,1] Take ALL rows, but only column index 1
```

```python
from sklearn.metrics import f1_score
f1_macro=f1_score(y_test,grid_pred,average='weighted')
print(f"f1_macro value for best parameter {rscore}:{f1_macro}")
```
